# Data Loading and Validation

```python
import pandas as pd
import numpy as np
from pathlib import Path

train_ground_truth = "D:/business_entity_resolution/dataset/train/train_ground_truth.tsv"
train_source1 = "D:/business_entity_resolution/dataset/train/train_source1.tsv"
train_source2 = "D:/business_entity_resolution/dataset/train/train_source2.tsv"
train_source3 = "D:/business_entity_resolution/dataset/train/train_source3.tsv"

Required_columns = [
    "entity_id",
    "business_name",
    "business_address",
    "country",
]

def load_tsv(file_path):

    
    file_path = Path(file_path)

    if not file_path.exists():
        raise FileNotFoundError(
            f"File not found {file_path}"
        )

    if file_path.suffix.lower() != ".tsv":
        raise ValueError(
            f"Expected a .tsv file, got: {file_path}"
        )

    df = pd.read_csv(
        file_path,
        sep="\t",
        dtype=str,
        keep_default_na=False
    )

    missing_columns = [column for column in Required_columns if column not in df.columns]

    if missing_columns:
        raise ValueError(
            f"{file_path.name} is missing required columns: "
            f"{missing_columns}"
        )

    return df   
```

```python
source1 = load_tsv(train_source1)
source1.head()
```

```python
source2 = load_tsv(train_source2)
source2.head()
```

```python
source3 = load_tsv(train_source3)
source3.head()
```

```python
gound_truth = pd.read_csv(train_ground_truth,sep="\t",dtype=str)
gound_truth.head()
```

# Data Processing

```python
# Data type and missing values handling

import pandas as pd


TEXT_COLUMNS = [
    "business_name",
    "business_address",
    "country",
]


def handle_missing_values(df):

    if df["entity_id"].isna().any():
        raise ValueError(
            "entity_id contains missing values."
        )

    df["entity_id"] = df["entity_id"].astype(str)

    # Handle text columns
    for column in TEXT_COLUMNS:
        df[column] = (
            df[column]
            .fillna("")
            .astype(str)
        )

    return df

source1 = handle_missing_values(source1)
source2 = handle_missing_values(source2)
source3 = handle_missing_values(source3)
```

```python
# unicode normalization
#Converts equivalent Unicode representations into a consistent form while preserving the original language/script. It does not translate or transliterate the text

import unicodedata

TEXT_COLUMNS = [
    "business_name",
    "business_address",
    "country",
]


def normalize_unicode(text):
    return unicodedata.normalize("NFKC", text)


def normalize_unicode_columns(df):
    df = df.copy()

    for column in TEXT_COLUMNS:
        df[column] = df[column].apply(normalize_unicode)

    return df
    
source1 = normalize_unicode_columns(source1)
source2 = normalize_unicode_columns(source2)
source3 = normalize_unicode_columns(source3)
```

```python
# Case normalization
#Convert text to lowercase so that case differences don't create unnecessary differences in later string-based similarity features

TEXT_COLUMNS = [
    "business_name",
    "business_address",
    "country",
]


def normalize_case(df):
    """
    Convert text columns to lowercase.
    """

    df = df.copy()

    for column in TEXT_COLUMNS:
        df[column] = df[column].str.lower()

    return df

source1 = normalize_case(source1)
source2 = normalize_case(source2)
source3 = normalize_case(source3)
```

```python
# Whitespace normalization

import re


TEXT_COLUMNS = [
    "business_name",
    "business_address",
    "country",
]


def normalize_whitespace(df):
    """
    Normalize repeated whitespace and remove
    leading/trailing whitespace.
    """

    df = df.copy()

    for column in TEXT_COLUMNS:
        df[column] = (
            df[column]
            .str.replace(r"\s+", " ", regex=True)
            .str.strip()
        )

    return df

source1 = normalize_whitespace(source1)
source2 = normalize_whitespace(source2)
source3 = normalize_whitespace(source3)
```

```python
# Field specific normalization

import re


def normalize_business_name(text):
    """
    Conservative normalization for business names.
    Keeps meaningful characters such as &, -, / and numbers.
    """
    text = re.sub(r"[.,;:!?()\[\]{}\"']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


def normalize_business_address(text):
    """
    Conservative normalization for business addresses.
    Keeps numbers, hyphens and slashes because they may
    represent meaningful address components.
    """
    text = re.sub(r"[.,;:!?()\[\]{}\"']", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text


def normalize_country(text):
    """
    Light normalization for country names.
    """
    return text.strip()


def add_field_specific_normalized_columns(df):
    """
    Add field-specific normalized columns while
    preserving the original columns.
    """
    df = df.copy()

    df["business_name_normalized"] = (
        df["business_name"]
        .apply(normalize_business_name)
    )

    df["business_address_normalized"] = (
        df["business_address"]
        .apply(normalize_business_address)
    )

    df["country_normalized"] = (
        df["country"]
        .apply(normalize_country)
    )

    return df
```

```python

source1 = add_field_specific_normalized_columns(source1)
source2 = add_field_specific_normalized_columns(source2)
source3 = add_field_specific_normalized_columns(source3)
```

```python
source1.to_csv("source1_processed.csv")
source2.to_csv("source2_processed.csv")
source3.to_csv("source3_processed.csv")
```
